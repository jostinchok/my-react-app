```python
from bing_image_downloader import downloader
import os
import re
import shutil

queries = [
    "hand touching leaf close up",
    "hand holding flower close up",
    "person plucking plant close up",
    "hand touching tree bark close up",
    "hand holding plant outdoor",
    "person holding potted plant close up",
]

base_dir = "datasets/touching-plants"
final_dir = base_dir
prefix = "touching_plants"

os.makedirs(final_dir, exist_ok=True)

# Find highest existing number, e.g. touching_plants_0234.jpg
pattern = re.compile(rf"{prefix}_(\d+)\.(jpg|jpeg|png)$", re.IGNORECASE)

max_num = 0
for file in os.listdir(final_dir):
    match = pattern.match(file)
    if match:
        max_num = max(max_num, int(match.group(1)))

next_num = max_num + 1
print(f"Starting from: {prefix}_{next_num:04d}")

# Download into temporary query folders
temp_dir = os.path.join(base_dir, "_new_downloads")

for q in queries:
    downloader.download(
        q,
        limit=100,
        output_dir=temp_dir,
        adult_filter_off=True,
        force_replace=False,
        timeout=60
    )

# Move and rename new images only
for root, _, files in os.walk(temp_dir):
    for file in files:
        if file.lower().endswith((".jpg", ".jpeg", ".png")):
            old_path = os.path.join(root, file)
            ext = os.path.splitext(file)[1].lower()

            new_name = f"{prefix}_{next_num:04d}{ext}"
            new_path = os.path.join(final_dir, new_name)

            shutil.move(old_path, new_path)
            print(f"Saved: {new_name}")

            next_num += 1

print("Done. Existing images were not overwritten.")
```
